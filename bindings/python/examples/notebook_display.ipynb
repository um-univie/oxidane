# Oxidane in Jupyter

Build `bindings/python` with the notebook extra and select that environment as the kernel. The old `molecules` import remains supported.

```python
from oxidane import Molecule, Smarts, Reaction, ReactionSmarts

ethanol = Molecule("CCO")
ethanol
```

```python
alcohol = Smarts("[OX2H1][CX4]")
assert ethanol.contains(alcohol)
assert ethanol.contains_smarts("[OX2H1]")
for smiles in ["CCO", "CC(=O)C"]:
    Molecule(smiles).plot()
```

```python
reaction = Reaction(["CCO"], ["CC=O"], name="Oxidation")
reaction.plot()
```

```python
rule = ReactionSmarts("[C:1]-[O:2]>>[C:1]=[O:2]")
rule
```

```python
rule.plot(reactants=[ethanol])
products = rule.apply(ethanol)
assert any(product.contains("C=O") for product in products)
```

```python
from molecules import PyMolecule
assert PyMolecule is Molecule
svg = reaction.to_svg(width=1000, height=300)
assert svg.startswith("<svg")
```
